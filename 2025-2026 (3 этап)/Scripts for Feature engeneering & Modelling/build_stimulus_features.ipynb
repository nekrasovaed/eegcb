```python
from glob import glob
import pandas as pd

files = glob('./Main/*/Text*')
files
```

```python
stimulus_features = pd.DataFrame()
for f in files:
    id = int(f.split('\\')[1][5:])
    temp = pd.read_csv(f)
    temp.index = [f"{id}_{i}" for i in temp['TextID']]
    #temp = temp.drop(columns=['TextID'])
    stimulus_features = pd.concat([stimulus_features, temp])

stimulus_features.to_csv('datasets/stimulus_features.csv')    
```
