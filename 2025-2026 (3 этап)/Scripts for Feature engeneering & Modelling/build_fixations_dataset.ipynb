```python
import pandas as pd
from tqdm import tqdm
```

```python
import glob

paths_fixations = glob.glob(f"./Main/*/EYE/*.csv")
paths_ordering = glob.glob(f"./Main/*/*.csv")

dataset = pd.DataFrame()
for path in tqdm(paths_fixations, desc='extracting and concatenating...'):
    _, participant_id, _, file = path.split('\\')
    order, screen_type = file.split('_')
    order = int(order.replace('fixation', ''))
    screen_type = screen_type[:-4]
    ordering = pd.read_csv(f"./Main/{participant_id}/TextOrdering.csv")
    text_id = ordering.iloc[order-1,:]['TextID']
    valence = ordering.iloc[order-1,:]['Valence']
    veracity = ordering.iloc[order-1,:]['Veracity']
    answer = ordering.iloc[order-1,:]['Performance Rate1']
    
    data = pd.read_csv(path)
    data['participant_id'] = int(participant_id[5:])
    data['screen_order'] = order
    data['stimul_id'] = text_id
    data['type'] = screen_type
    data['Valence'] = valence
    data['Veracity'] = veracity
    data['answer'] = answer
    
    dataset = pd.concat([dataset, data], axis=0)
    
```

```python
dataset = dataset.sort_values(['participant_id', 'screen_order', 'type'], ascending=[True, True, False]).reset_index(drop=True)
dataset
```

```python
dataset.to_csv("datasets/main_fixations.csv")
```
